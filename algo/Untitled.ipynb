```python
import numpy as np
import random
import matplotlib.pyplot as plt

# Define the given 7x5 matrix (Power Flow Coefficients)
# Each row represents a line, each column represents a bus
A = np.array([
    [0.5857, -0.2571, -0.0428, -0.0857, -0.2000],
    [0.2143,  0.0571, -0.1572, -0.1143, -0.0000],
    [0.0905,  0.1619, -0.1953, -0.1238,  0.0666],
    [0.1080,  0.1651, -0.1206, -0.1968,  0.0444],
    [0.1872,  0.2158,  0.0730,  0.0349, -0.5109],
    [0.1048,  0.0191,  0.4476, -0.4380, -0.1334],
    [0.0128, -0.0158,  0.1270,  0.1651, -0.2891]
])

# User input for Line Power Limits (7 lines)
line_limits = np.array([float(input(f"Enter power limit for Line {i+1}: ")) for i in range(7)])

# User input for Bus Injection Matrix B (5x1)
B = np.array([float(input(f"Enter power injection for Bus {i+1}: ")) for i in range(5)]).reshape(5, 1)

# Hybrid Optimization Algorithm with Strict Constraint Enforcement
def hybrid_optimization(A, B, line_limits, iterations=100):
    num_buses = A.shape[1]
    best_solution = B.copy()
    best_fitness = np.inf
    fitness_history = []
    
    C_initial = np.dot(A, B)  # Compute initial power flow on lines
    total_power_flow_initial = np.sum(abs(C_initial))  # Store total initial power flow
    
    for iteration in range(iterations):
        candidate_solution = best_solution.copy()
        C = np.dot(A, candidate_solution)
        
        # Identify overloaded lines
        overloaded_lines = np.where(abs(C.flatten()) > line_limits)[0]
        
        if len(overloaded_lines) > 0:
            # Calculate sensitivity matrix (how each bus injection affects each line flow)
            sensitivity = A[overloaded_lines, :]
            
            # Calculate required adjustment for overloaded lines
            excess = abs(C[overloaded_lines]) - line_limits[overloaded_lines]
            
            # Adjust injections to reduce overloaded flows
            for i, line in enumerate(overloaded_lines):
                # Find which bus injections have the most impact on this line
                impact_factors = sensitivity[i, :]
                highest_impact_idx = np.argsort(-abs(impact_factors))  # Sort by impact
                
                # Adjust the injection with highest impact
                for idx in highest_impact_idx:
                    if impact_factors[idx] != 0:
                        # Direction of adjustment depends on sign of sensitivity
                        adjustment = -excess[i] * np.sign(impact_factors[idx]) * 0.5
                        candidate_solution[idx] += adjustment
                        break
        
        # Ensure the sum of injections remains zero (power balance)
        candidate_solution -= np.mean(candidate_solution)
        
        # Recalculate power flows after adjustment
        C_new = np.dot(A, candidate_solution)
        
        # Calculate fitness with penalty for constraint violations
        penalty = np.sum(np.maximum(0, abs(C_new.flatten()) - line_limits))
        fitness = np.linalg.norm(C_new) + penalty * 10  # Increased penalty weight
        fitness_history.append(fitness)
        
        if fitness < best_fitness:
            best_fitness = fitness
            best_solution = candidate_solution.copy()
            
        # Early stopping if we've found a good solution
        if penalty == 0 and iteration > 20:
            break
    
    # Final correction to enforce sum(B_optimized) = 0
    best_solution -= np.mean(best_solution)
    
    # Calculate final power flows
    C_optimized = np.dot(A, best_solution)
    
    return best_solution, fitness_history, C_optimized

# Compute initial (unoptimized) power flows
C_unoptimized = np.dot(A, B)

# Run the Hybrid Algorithm
B_optimized, fitness_history, C_optimized = hybrid_optimization(A, B, line_limits)

# Plot results
plt.figure(figsize=(10, 5))
plt.plot(range(1, 8), C_unoptimized, marker='o', linestyle='-', label='Unoptimized Line Flows')
plt.plot(range(1, 8), C_optimized, marker='s', linestyle='--', label='Optimized Line Flows')

# Plot line limits
for i in range(7):
    plt.axhline(y=line_limits[i], color='g', linestyle=':', alpha=0.5)
    plt.axhline(y=-line_limits[i], color='g', linestyle=':', alpha=0.5)

plt.axhline(y=line_limits.mean(), color='r', linestyle='dotted', label='Average Line Limit')
plt.xlabel('Line Index')
plt.ylabel('Power Flow')
plt.title('Power Flow Comparison Before and After Optimization')
plt.legend()
plt.grid()
plt.show()

# Plot fitness score history
plt.figure(figsize=(10, 5))
plt.plot(range(len(fitness_history)), fitness_history, marker='o', linestyle='-', color='b', label='Fitness Score')
plt.xlabel('Iteration')
plt.ylabel('Fitness Score')
plt.title('Fitness Score Convergence Over Iterations')
plt.legend()
plt.grid()
plt.show()

# Plot power injection comparison
plt.figure(figsize=(10, 5))
indices = range(1, 6)
plt.bar(indices, B.flatten(), width=0.4, label='Initial Injections', alpha=0.7, color='blue')
plt.bar([i+0.4 for i in indices], B_optimized.flatten(), width=0.4, label='Optimized Injections', alpha=0.7, color='green')
plt.xlabel('Bus Index')
plt.ylabel('Power Injection')
plt.title('Bus Power Injection Comparison')
plt.legend()
plt.grid(axis='y')
plt.show()

# Print results
print("\nMatrix A (7x5):\n", A)
print("\nInitial Power Injections (B, 5x1):\n", B)
print("\nResultant Line Flows Before Optimization (7x1):\n", C_unoptimized)
print("\nOptimized Power Injections (B_optimized, 5x1):\n", B_optimized)
print("\nSum of Optimized Power Injections (Should be 0): ", np.sum(B_optimized))
print("\nResultant Line Flows After Optimization (7x1):\n", C_optimized)

# Check that constraints are satisfied
violations = np.where(abs(C_optimized.flatten()) > line_limits)[0]
if len(violations) > 0:
    print(f"\nWarning: {len(violations)} line constraints still violated!")
    for line in violations:
        print(f"  Line {line+1}: Flow = {abs(C_optimized[line][0]):.4f}, Limit = {line_limits[line]:.4f}")
else:
    print("\nAll line constraints satisfied!")

# Check total power flow
print("\nTotal Line Power Flow Before Optimization:", np.sum(abs(C_unoptimized)))
print("Total Line Power Flow After Optimization:", np.sum(abs(C_optimized)))
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt

# Define the given 7x5 matrix (Power Flow Coefficients)
# Each row represents a line, each column represents a bus
A = np.array([
    [0.5857, -0.2571, -0.0428, -0.0857, -0.2000],
    [0.2143,  0.0571, -0.1572, -0.1143, -0.0000],
    [0.0905,  0.1619, -0.1953, -0.1238,  0.0666],
    [0.1080,  0.1651, -0.1206, -0.1968,  0.0444],
    [0.1872,  0.2158,  0.0730,  0.0349, -0.5109],
    [0.1048,  0.0191,  0.4476, -0.4380, -0.1334],
    [0.0128, -0.0158,  0.1270,  0.1651, -0.2891]
])

# User input for Line Power Limits (7 lines)
line_limits = np.array([float(input(f"Enter power limit for Line {i+1}: ")) for i in range(7)])

# User input for Bus Injection Matrix B (5x1)
B = np.array([float(input(f"Enter power injection for Bus {i+1}: ")) for i in range(5)]).reshape(5, 1)

# Hybrid Optimization Algorithm with Strict Constraint Enforcement and Flow Preservation
def hybrid_optimization(A, B, line_limits, iterations=200, max_flow_change_pct=2.0):
    num_buses = A.shape[1]
    best_solution = B.copy()
    best_fitness = np.inf
    fitness_history = []
    
    C_initial = np.dot(A, B)  # Compute initial power flow on lines
    total_power_flow_initial = np.sum(abs(C_initial))  # Store total initial power flow
    
    # Define acceptable range for total power flow
    min_total_flow = total_power_flow_initial * (1 - max_flow_change_pct/100)
    max_total_flow = total_power_flow_initial * (1 + max_flow_change_pct/100)
    
    for iteration in range(iterations):
        candidate_solution = best_solution.copy()
        C = np.dot(A, candidate_solution)
        
        # Identify overloaded lines
        overloaded_lines = np.where(abs(C.flatten()) > line_limits)[0]
        
        if len(overloaded_lines) > 0:
            # Calculate sensitivity matrix (how each bus injection affects each line flow)
            sensitivity = A[overloaded_lines, :]
            
            # Calculate required adjustment for overloaded lines
            excess = abs(C[overloaded_lines]) - line_limits[overloaded_lines]
            
            # Adjust injections to reduce overloaded flows
            for i, line in enumerate(overloaded_lines):
                # Find which bus injections have the most impact on this line
                impact_factors = sensitivity[i, :]
                highest_impact_idx = np.argsort(-abs(impact_factors))  # Sort by impact
                
                # Adjust the injection with highest impact
                for idx in highest_impact_idx[:2]:  # Try top 2 impact factors
                    if impact_factors[idx] != 0:
                        # Direction of adjustment depends on sign of sensitivity and flow
                        flow_sign = np.sign(C[line][0])
                        impact_sign = np.sign(impact_factors[idx])
                        
                        # If impact_sign and flow_sign are the same, then reducing injection reduces flow
                        if impact_sign * flow_sign > 0:
                            adjustment = -excess[i] * 0.7  # More aggressive adjustment
                        else:
                            adjustment = excess[i] * 0.7
                            
                        candidate_solution[idx] += adjustment * impact_sign
        
        # Ensure the sum of injections remains zero (power balance)
        candidate_solution -= np.mean(candidate_solution)
        
        # Recalculate power flows after adjustment
        C_new = np.dot(A, candidate_solution)
        total_flow_new = np.sum(abs(C_new))
        
        # Calculate flow change penalty
        flow_change_pct = abs((total_flow_new - total_power_flow_initial) / total_power_flow_initial * 100)
        flow_change_penalty = max(0, flow_change_pct - max_flow_change_pct) * 50  # High penalty if exceeding threshold
        
        # Calculate fitness with penalty for constraint violations and flow changes
        constraint_penalty = np.sum(np.maximum(0, abs(C_new.flatten()) - line_limits))
        fitness = constraint_penalty * 100 + np.sum(abs(C_new - C_initial)) + flow_change_penalty
        fitness_history.append(fitness)
        
        if fitness < best_fitness:
            best_fitness = fitness
            best_solution = candidate_solution.copy()
            
        # Early stopping if we've found a solution with no violations and within flow change limits
        if constraint_penalty == 0 and flow_change_pct <= max_flow_change_pct and iteration > 50:
            break
    
    # Final correction to enforce sum(B_optimized) = 0
    best_solution -= np.mean(best_solution)
    
    # Calculate final power flows
    C_optimized = np.dot(A, best_solution)
    
    # Final constraint check and adjustment if needed
    max_iterations = 50
    iteration = 0
    
    while iteration < max_iterations:
        violations = np.where(abs(C_optimized.flatten()) > line_limits)[0]
        if len(violations) == 0:
            break
            
        # Find most overloaded line
        most_overloaded = violations[np.argmax(abs(C_optimized[violations]) - line_limits[violations])]
        excess = abs(C_optimized[most_overloaded][0]) - line_limits[most_overloaded]
        
        # Find which bus has most impact on this line
        impact_factors = A[most_overloaded, :]
        highest_impact_idx = np.argmax(abs(impact_factors))
        
        # Adjust the injection with highest impact
        flow_sign = np.sign(C_optimized[most_overloaded][0])
        impact_sign = np.sign(impact_factors[highest_impact_idx])
        
        if impact_sign * flow_sign > 0:
            adjustment = -excess * 1.1  # Slightly over-adjust to ensure constraint is met
        else:
            adjustment = excess * 1.1
            
        best_solution[highest_impact_idx] += adjustment * impact_sign
        
        # Rebalance
        best_solution -= np.mean(best_solution)
        
        # Recalculate
        C_optimized = np.dot(A, best_solution)
        iteration += 1
    
    # Enhanced flow preservation
    # Calculate current total flow
    total_power_flow_optimized = np.sum(abs(C_optimized))
    flow_change_pct = (total_power_flow_optimized - total_power_flow_initial) / total_power_flow_initial * 100
    
    # If flow difference exceeds target, try to adjust
    if abs(flow_change_pct) > max_flow_change_pct:
        # Determine target total flow within acceptable range
        if flow_change_pct > max_flow_change_pct:
            target_flow = max_total_flow
        else:
            target_flow = min_total_flow
            
        target_ratio = target_flow / total_power_flow_optimized
        
        # Try multiple scale factors in small increments to find one that meets constraints
        # and gets closer to target flow
        best_scale_factor = 1.0
        closest_flow_diff = abs(flow_change_pct)
        
        for scale_factor in np.linspace(1.0, target_ratio, 20):
            test_solution = best_solution * scale_factor
            test_solution -= np.mean(test_solution)  # Rebalance
            test_flows = np.dot(A, test_solution)
            
            # Check if constraints are violated
            if np.any(abs(test_flows) > line_limits):
                continue
            
            # Calculate flow difference
            test_total_flow = np.sum(abs(test_flows))
            test_flow_change_pct = (test_total_flow - total_power_flow_initial) / total_power_flow_initial * 100
            
            # If this is better than our previous best, update
            if abs(test_flow_change_pct) < closest_flow_diff:
                closest_flow_diff = abs(test_flow_change_pct)
                best_scale_factor = scale_factor
                
                # If we're within tolerance, we can stop
                if abs(test_flow_change_pct) <= max_flow_change_pct:
                    break
        
        # Apply the best scale factor
        best_solution = best_solution * best_scale_factor
        best_solution -= np.mean(best_solution)  # Final balance
    
    # Calculate final optimized flows
    C_optimized = np.dot(A, best_solution)
    
    return best_solution, fitness_history, C_optimized

# Compute initial (unoptimized) power flows
C_unoptimized = np.dot(A, B)

# Run the Hybrid Algorithm with stricter flow preservation
B_optimized, fitness_history, C_optimized = hybrid_optimization(A, B, line_limits, max_flow_change_pct=2.0)

# Plot results
plt.figure(figsize=(10, 5))
plt.plot(range(1, 8), C_unoptimized, marker='o', linestyle='-', label='Unoptimized Line Flows')
plt.plot(range(1, 8), C_optimized, marker='s', linestyle='--', label='Optimized Line Flows')

# Plot line limits
for i in range(7):
    plt.axhline(y=line_limits[i], color='g', linestyle=':', alpha=0.5)
    plt.axhline(y=-line_limits[i], color='g', linestyle=':', alpha=0.5)

plt.axhline(y=line_limits.mean(), color='r', linestyle='dotted', label='Average Line Limit')
plt.xlabel('Line Index')
plt.ylabel('Power Flow')
plt.title('Power Flow Comparison Before and After Optimization')
plt.legend()
plt.grid()
plt.show()

# Plot fitness score history
plt.figure(figsize=(10, 5))
plt.plot(range(len(fitness_history)), fitness_history, marker='o', linestyle='-', color='b', label='Fitness Score')
plt.xlabel('Iteration')
plt.ylabel('Fitness Score')
plt.title('Fitness Score Convergence Over Iterations')
plt.legend()
plt.grid()
plt.show()

# Plot power injection comparison
plt.figure(figsize=(10, 5))
indices = range(1, 6)
plt.bar(indices, B.flatten(), width=0.4, label='Initial Injections', alpha=0.7, color='blue')
plt.bar([i+0.4 for i in indices], B_optimized.flatten(), width=0.4, label='Optimized Injections', alpha=0.7, color='green')
plt.xlabel('Bus Index')
plt.ylabel('Power Injection')
plt.title('Bus Power Injection Comparison')
plt.legend()
plt.grid(axis='y')
plt.show()

# Plot total power flow comparison
plt.figure(figsize=(8, 4))
total_flow_before = np.sum(abs(C_unoptimized))
total_flow_after = np.sum(abs(C_optimized))
plt.bar(['Before Optimization', 'After Optimization'], 
       [total_flow_before, total_flow_after], 
       color=['blue', 'green'])
plt.ylabel('Total Absolute Line Flow')
plt.title('Total Power Flow Comparison')
plt.grid(axis='y')
plt.show()

# Print results
print("\nMatrix A (7x5):\n", A)
print("\nInitial Power Injections (B, 5x1):\n", B)
print("\nResultant Line Flows Before Optimization (7x1):\n", C_unoptimized)
print("\nOptimized Power Injections (B_optimized, 5x1):\n", B_optimized)
print("\nSum of Optimized Power Injections (Should be 0): ", np.sum(B_optimized))
print("\nResultant Line Flows After Optimization (7x1):\n", C_optimized)

# Check that constraints are satisfied
violations = np.where(abs(C_optimized.flatten()) > line_limits)[0]
if len(violations) > 0:
    print(f"\nWarning: {len(violations)} line constraints still violated!")
    for line in violations:
        print(f"  Line {line+1}: Flow = {abs(C_optimized[line][0]):.4f}, Limit = {line_limits[line]:.4f}")
else:
    print("\nAll line constraints satisfied!")

# Calculate flow preservation metrics
print("\nTotal Line Power Flow Before Optimization:", np.sum(abs(C_unoptimized)))
print("Total Line Power Flow After Optimization:", np.sum(abs(C_optimized)))
flow_diff = np.sum(abs(C_optimized)) - np.sum(abs(C_unoptimized))
flow_pct = (flow_diff/np.sum(abs(C_unoptimized)))*100
print(f"Difference: {flow_diff:.6f}")
print(f"Percentage Difference: {flow_pct:.6f}%")

# Check if flow difference is within target range
if abs(flow_pct) <= 2.0:
    print("\n✓ Total flow difference is within the ±2% target range")
else:
    print("\n⚠ Total flow difference exceeds the ±2% target range")

print("\nConstraint verification:")
for i in range(7):
    margin = line_limits[i] - abs(C_optimized[i][0])
    print(f"  Line {i+1}: Flow = {abs(C_optimized[i][0]):.4f}, Limit = {line_limits[i]:.4f}, Margin = {margin:.4f}")
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt

# Define the given 7x5 matrix (Power Flow Coefficients)
# Each row represents a line, each column represents a bus
A = np.array([
    [0.5857, -0.2571, -0.0428, -0.0857, -0.2000],
    [0.2143,  0.0571, -0.1572, -0.1143, -0.0000],
    [0.0905,  0.1619, -0.1953, -0.1238,  0.0666],
    [0.1080,  0.1651, -0.1206, -0.1968,  0.0444],
    [0.1872,  0.2158,  0.0730,  0.0349, -0.5109],
    [0.1048,  0.0191,  0.4476, -0.4380, -0.1334],
    [0.0128, -0.0158,  0.1270,  0.1651, -0.2891]
])

# User input for Line Power Limits (7 lines)
line_limits = np.array([float(input(f"Enter power limit for Line {i+1}: ")) for i in range(7)])

# User input for Bus Injection Matrix B (5x1)
B = np.array([float(input(f"Enter power injection for Bus {i+1}: ")) for i in range(5)]).reshape(5, 1)

# Targeted Optimization Algorithm with Minimal Flow Disruption
def targeted_optimization(A, B, line_limits, iterations=250, max_flow_change_pct=2.0):
    num_buses = A.shape[1]
    num_lines = A.shape[0]
    best_solution = B.copy()
    best_fitness = np.inf
    fitness_history = []
    
    C_initial = np.dot(A, B)  # Compute initial power flow on lines
    total_power_flow_initial = np.sum(abs(C_initial))  # Store total initial power flow
    
    # Identify initially overloaded lines
    overloaded_lines = np.where(abs(C_initial.flatten()) > line_limits)[0]
    print(f"Initially overloaded lines: {[i+1 for i in overloaded_lines]}")
    
    # If no lines are overloaded, return original solution
    if len(overloaded_lines) == 0:
        return B, [0], C_initial
    
    # Create a mask for lines we want to maintain as close as possible to original
    non_overloaded_mask = np.ones(num_lines, dtype=bool)
    non_overloaded_mask[overloaded_lines] = False
    
    for iteration in range(iterations):
        candidate_solution = best_solution.copy()
        C = np.dot(A, candidate_solution)
        
        # Identify currently overloaded lines
        current_overloaded = np.where(abs(C.flatten()) > line_limits)[0]
        
        if len(current_overloaded) > 0:
            # Extract sensitivity submatrix for overloaded lines
            sensitivity = A[current_overloaded, :]
            
            # Calculate required adjustment for overloaded lines
            excess = abs(C[current_overloaded]) - line_limits[current_overloaded]
            
            # Adjust injections to reduce overloaded flows
            for i, line in enumerate(current_overloaded):
                # Find which bus injections have the most impact on this line
                impact_factors = sensitivity[i, :]
                highest_impact_idx = np.argsort(-abs(impact_factors))
                
                # Calculate how changes will affect other lines
                for idx in highest_impact_idx[:2]:  # Try top 2 impact factors
                    if impact_factors[idx] != 0:
                        # Direction of adjustment depends on sign of sensitivity and flow
                        flow_sign = np.sign(C[line][0])
                        impact_sign = np.sign(impact_factors[idx])
                        
                        # Calculate adjustment amount
                        if impact_sign * flow_sign > 0:
                            adjustment = -excess[i] * 0.8
                        else:
                            adjustment = excess[i] * 0.8
                        
                        # Predict effect on non-overloaded lines
                        test_solution = candidate_solution.copy()
                        test_solution[idx] += adjustment * impact_sign
                        test_solution -= np.mean(test_solution)  # Rebalance
                        test_flows = np.dot(A, test_solution)
                        
                        # Check if adjustment creates new overloads
                        new_overloads = False
                        for j in np.where(non_overloaded_mask)[0]:
                            if abs(test_flows[j][0]) > line_limits[j]:
                                new_overloads = True
                                break
                        
                        # If no new overloads, apply adjustment
                        if not new_overloads:
                            candidate_solution[idx] += adjustment * impact_sign
                            break
        
        # Ensure the sum of injections remains zero (power balance)
        candidate_solution -= np.mean(candidate_solution)
        
        # Recalculate power flows after adjustment
        C_new = np.dot(A, candidate_solution)
        
        # Calculate fitness with focus on targeted lines and minimizing changes
        # 1. Major penalty for any overloaded line
        overload_penalty = np.sum(np.maximum(0, abs(C_new.flatten()) - line_limits))
        
        # 2. Penalty for changing non-overloaded lines
        flow_preservation_penalty = np.sum(abs(C_new[non_overloaded_mask] - C_initial[non_overloaded_mask]))
        
        # 3. Check total flow change
        total_flow_new = np.sum(abs(C_new))
        flow_change_pct = abs((total_flow_new - total_power_flow_initial) / total_power_flow_initial * 100)
        flow_total_penalty = max(0, flow_change_pct - max_flow_change_pct) * 20
        
        # Combined fitness
        fitness = overload_penalty * 100 + flow_preservation_penalty * 10 + flow_total_penalty
        fitness_history.append(fitness)
        
        if fitness < best_fitness:
            best_fitness = fitness
            best_solution = candidate_solution.copy()
            
        # Early stopping if constraints satisfied and minimal disruption
        if overload_penalty == 0 and flow_change_pct <= max_flow_change_pct and iteration > 20:
            break
    
    # Final correction to ensure power balance
    best_solution -= np.mean(best_solution)
    
    # Calculate final power flows
    C_optimized = np.dot(A, best_solution)
    
    # Final check for any remaining overloaded lines
    violated_lines = np.where(abs(C_optimized.flatten()) > line_limits)[0]
    iteration = 0
    max_iterations = 30
    
    # If there are still violations, use targeted adjustments on individual lines
    while len(violated_lines) > 0 and iteration < max_iterations:
        # Focus on most overloaded line first
        line_idx = violated_lines[np.argmax(abs(C_optimized[violated_lines]) - line_limits[violated_lines])]
        excess = abs(C_optimized[line_idx][0]) - line_limits[line_idx]
        
        # Find buses with highest impact on this line
        impact_factors = A[line_idx, :]
        highest_impact_idx = np.argmax(abs(impact_factors))
        
        # Calculate adjustment needed
        flow_sign = np.sign(C_optimized[line_idx][0])
        impact_sign = np.sign(impact_factors[highest_impact_idx])
        
        if impact_sign * flow_sign > 0:
            adjustment = -excess * 1.05
        else:
            adjustment = excess * 1.05
            
        # Apply adjustment
        best_solution[highest_impact_idx] += adjustment * impact_sign
        
        # Rebalance
        best_solution -= np.mean(best_solution)
        
        # Recalculate flows
        C_optimized = np.dot(A, best_solution)
        
        # Check for violations again
        violated_lines = np.where(abs(C_optimized.flatten()) > line_limits)[0]
        iteration += 1
    
    # Flow preservation final step - try to minimize changes to non-overloaded lines
    total_power_flow_optimized = np.sum(abs(C_optimized))
    flow_change_pct = (total_power_flow_optimized - total_power_flow_initial) / total_power_flow_initial * 100
    
    # Only adjust if needed and won't violate constraints
    if abs(flow_change_pct) > max_flow_change_pct:
        target_total_flow = total_power_flow_initial * (1 + np.sign(flow_change_pct) * max_flow_change_pct/100)
        current_ratio = total_power_flow_optimized / total_power_flow_initial
        
        # Try progressive scaling factors
        for scale_factor in np.linspace(1.0, target_total_flow/total_power_flow_optimized, 15):
            test_solution = best_solution * scale_factor
            test_solution -= np.mean(test_solution)  # Rebalance
            test_flows = np.dot(A, test_solution)
            
            # Check if constraints still satisfied
            if not np.any(abs(test_flows) > line_limits):
                best_solution = test_solution.copy()
                C_optimized = test_flows.copy()
                break
    
    return best_solution, fitness_history, C_optimized

# Compute initial (unoptimized) power flows
C_unoptimized = np.dot(A, B)

# Run the Targeted Optimization Algorithm
B_optimized, fitness_history, C_optimized = targeted_optimization(A, B, line_limits, max_flow_change_pct=2.0)

# Plot results
plt.figure(figsize=(12, 6))
plt.plot(range(1, 8), C_unoptimized, marker='o', linestyle='-', linewidth=2, label='Unoptimized Line Flows')
plt.plot(range(1, 8), C_optimized, marker='s', linestyle='--', linewidth=2, label='Optimized Line Flows')

# Plot line limits
for i in range(7):
    plt.axhline(y=line_limits[i], color='g', linestyle=':', alpha=0.5)
    plt.axhline(y=-line_limits[i], color='g', linestyle=':', alpha=0.5)

plt.axhline(y=line_limits.mean(), color='r', linestyle='dotted', label='Average Line Limit')
plt.xlabel('Line Index')
plt.ylabel('Power Flow')
plt.title('Power Flow Comparison Before and After Optimization')
plt.legend()
plt.grid(True)
plt.show()

# Plot fitness score history
if len(fitness_history) > 1:
    plt.figure(figsize=(10, 5))
    plt.plot(range(len(fitness_history)), fitness_history, marker='o', linestyle='-', color='b', label='Fitness Score')
    plt.xlabel('Iteration')
    plt.ylabel('Fitness Score')
    plt.title('Fitness Score Convergence Over Iterations')
    plt.legend()
    plt.grid(True)
    plt.show()

# Plot power injection comparison
plt.figure(figsize=(10, 5))
indices = range(1, 6)
plt.bar(indices, B.flatten(), width=0.4, label='Initial Injections', alpha=0.7, color='blue')
plt.bar([i+0.4 for i in indices], B_optimized.flatten(), width=0.4, label='Optimized Injections', alpha=0.7, color='green')
plt.xlabel('Bus Index')
plt.ylabel('Power Injection')
plt.title('Bus Power Injection Comparison')
plt.legend()
plt.grid(axis='y')
plt.show()

# Plot flow change by line
plt.figure(figsize=(12, 6))
line_indices = range(1, 8)
flow_changes = [(abs(C_optimized[i][0]) - abs(C_unoptimized[i][0])) for i in range(7)]
colors = ['red' if abs(C_unoptimized[i][0]) > line_limits[i] else 'blue' for i in range(7)]
plt.bar(line_indices, flow_changes, color=colors)
plt.axhline(y=0, color='black', linestyle='-', linewidth=1)
plt.xlabel('Line Index')
plt.ylabel('Flow Change')
plt.title('Flow Change by Line (Red = Initially Overloaded)')
plt.grid(axis='y')
plt.show()

# Plot total power flow comparison
plt.figure(figsize=(8, 4))
total_flow_before = np.sum(abs(C_unoptimized))
total_flow_after = np.sum(abs(C_optimized))
plt.bar(['Before Optimization', 'After Optimization'], 
       [total_flow_before, total_flow_after], 
       color=['blue', 'green'])
plt.ylabel('Total Absolute Line Flow')
plt.title('Total Power Flow Comparison')
plt.grid(axis='y')
plt.show()

# Print results
print("\nMatrix A (7x5):\n", A)
print("\nInitial Power Injections (B, 5x1):\n", B)
print("\nResultant Line Flows Before Optimization (7x1):\n", C_unoptimized)
print("\nOptimized Power Injections (B_optimized, 5x1):\n", B_optimized)
print("\nSum of Optimized Power Injections (Should be 0): ", np.sum(B_optimized))
print("\nResultant Line Flows After Optimization (7x1):\n", C_optimized)

# Check that constraints are satisfied
violations = np.where(abs(C_optimized.flatten()) > line_limits)[0]
if len(violations) > 0:
    print(f"\nWarning: {len(violations)} line constraints still violated!")
    for line in violations:
        print(f"  Line {line+1}: Flow = {abs(C_optimized[line][0]):.4f}, Limit = {line_limits[line]:.4f}")
else:
    print("\nAll line constraints satisfied!")

# Calculate flow preservation metrics
print("\nTotal Line Power Flow Before Optimization:", np.sum(abs(C_unoptimized)))
print("Total Line Power Flow After Optimization:", np.sum(abs(C_optimized)))
flow_diff = np.sum(abs(C_optimized)) - np.sum(abs(C_unoptimized))
flow_pct = (flow_diff/np.sum(abs(C_unoptimized)))*100
print(f"Difference: {flow_diff:.6f}")
print(f"Percentage Difference: {flow_pct:.6f}%")

# Check if flow difference is within target range
if abs(flow_pct) <= 2.0:
    print("\n✓ Total flow difference is within the ±2% target range")
else:
    print("\n⚠ Total flow difference exceeds the ±2% target range")

# Line-by-line flow change analysis
print("\nLine-by-line flow changes:")
print("Line  |  Before  |  After  |  Change  |  % Change  |  Limit  |  Status")
print("-" * 70)
for i in range(7):
    flow_before = abs(C_unoptimized[i][0])
    flow_after = abs(C_optimized[i][0])
    change = flow_after - flow_before
    pct_change = (change/flow_before)*100 if flow_before != 0 else float('inf')
    
    if flow_before > line_limits[i]:
        status = "FIXED"
    elif abs(pct_change) < 1.0:
        status = "PRESERVED"
    else:
        status = "MODIFIED"
        
    print(f"{i+1:4d}  |  {flow_before:7.4f}  |  {flow_after:7.4f}  |  {change:8.4f}  |  {pct_change:10.2f}%  |  {line_limits[i]:6.4f}  |  {status}")

print("\nConstraint verification:")
for i in range(7):
    margin = line_limits[i] - abs(C_optimized[i][0])
    print(f"  Line {i+1}: Flow = {abs(C_optimized[i][0]):.4f}, Limit = {line_limits[i]:.4f}, Margin = {margin:.4f}")
```
